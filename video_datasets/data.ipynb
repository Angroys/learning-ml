```python
fig, axs = plt.subplots(20, 4, figsize=(50, 200)) # 2 rows, 2 columns
r = 0
c = 0 
for i in stress.columns[0:-2]:
    for j in stress.columns[0:-2]:
        if i !=j:
            title = i + " + " + j 
            axs[r, c].scatter(stress[i], stress[j] )
            axs[r, c].set_title(title)
            c +=1
            if c == 4:
                c= 0
    r+=1
    if r == 20:
        r = 0

plt.show()
```

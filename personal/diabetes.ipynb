```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("diabetes_dataset.csv")
```

```python
df
```

```python
y_train = np.array(df["Outcome"][0:50])
y_train =y_train.reshape(-1, 1)
```

```python
def plot_data(X,y ,ax, pos_label="y=1" ,neg_label = "y=0", s=80, loc ="best"):
    pos = y ==1
    neg = y ==0
    pos = pos.reshape(-1,) 
    neg = neg.reshape(-1,)
    ax.scatter(X[pos, 0], X[pos, 1], marker ="x", s=s, c='red', label=pos_label)
    ax.scatter(X[neg, 0], X[neg, 1], marker ="o", s=s, label=neg_label, facecolors='none', edgecolors="blue", lw=3)
```

```python
column_names = list(df.columns.values)
column_names = column_names[0:-1]
column_names

```

```python
fig,ax = plt.subplots(16,5,figsize=(50,70))

i, k = 0, 0
num = 0

while(i < 15):
    for j in range(1, 14 -num):
        x_train = np.array([df[column_names[num]][0:50],df[column_names[j]][0:50] ])
        x_train = np.column_stack((x_train[0], x_train[1]))
        plot_data(x_train, y_train, ax[i][k])
        ax[i][k].set_title(rf"{column_names[num]}  {column_names[j]}")
        if k == 4:
            k = 0
            i+=1
        else:
            k += 1
    num +=1


        

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
def calculate_sigmoid_values(z):
    g = 1/(1+np.exp(-z))

    return g
```

```python
z = np.arange(-10, 11)

y = calculate_sigmoid_values(z)
np.set_printoptions(precision=3) 
print(np.c_[z, y])
```

```python
fix, ax = plt.subplots(1,1, figsize=(12,4))

ax.plot(z, y, c="b")
ax.set_xlabel("z")
ax.set_ylabel("sigmoid(z)")
```

# Decision Boundary


```python
def plot_data(X,y ,ax, pos_label="y=1" ,neg_label = "y=0", s=80, loc ="best"):
    pos = y ==1
    neg = y ==0
    pos = pos.reshape(-1,) 
    neg = neg.reshape(-1,)
    ax.scatter(X[pos, 0], X[pos, 1], marker ="x", s=s, c='red', label=pos_label)
    ax.scatter(X[neg, 0], X[neg, 1], marker ="o", s=s, label=neg_label, facecolors='none', edgecolors="blue", lw=3)
```

```python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1]).reshape(-1,1) 

fig,ax = plt.subplots(1,1,figsize=(4,4))
plot_data(X, y, ax)
ax.set_title("Decision Boundary")

ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.show()
```

```python
z = np.arange(-10,11,)

fig,ax = plt.subplots(1,1,figsize=(5,3))
# Plot z vs sigmoid(z)
ax.plot(z, calculate_sigmoid_values(z), c="b")
ax.fill_between([-10, 0],[1, 1], alpha = 0.3)
ax.fill_between([10, 0],[1, 1], color= "red", alpha=0.3)
ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
```

```python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1]).reshape(-1,1) 

fig,ax = plt.subplots(1,1,figsize=(4,4))
ax.fill_between([0,3], [3, 0], alpha=0.2)
plot_data(X, y, ax)
ax.set_title("Decision Boundary")

ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.show()
```
